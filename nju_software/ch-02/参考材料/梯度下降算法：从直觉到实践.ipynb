# 梯度下降算法实践：房价预测

本文档基于《梯度下降算法：从直觉到实践》文档中的 1.5 节示例生成，旨在通过代码复现梯度下降的过程并验证计算结果。

## 1. 问题背景

根据房屋面积预测价格：

| 面积（㎡） | 价格（万元） |
| ---------- | ------------ |
| 50         | 150          |
| 80         | 240          |
| 100        | 300          |
| 120        | 360          |
| 150        | 450          |


```python
import numpy as np
import pandas as pd

# 准备数据
x = np.array([50, 80, 100, 120, 150])
y = np.array([150, 240, 300, 360, 450])
n = len(x)

print("数据准备完成。")
print(f"x (面积): {x}")
print(f"y (价格): {y}")
```

## 2. 梯度下降实现

我们将实现梯度下降算法来寻找最佳的 $w$ 和 $b$。

**参数设置**：
- 初始参数：$w=0, b=0$
- 学习率：$\eta = 1e-6$
- 迭代次数：5000

```python
# 初始化参数
w = 0.0
b = 0.0
lr = 1e-6  # 学习率
epochs = 5000

# 指定需要打印日志的迭代轮次
log_epochs = [0, 1, 10, 50, 100, 500, 1000, 2000, 5000]

print(f"{'Iteration':<10} {'w':<10} {'b':<10} {'Loss':<15} {'dJ/dw':<15} {'dJ/db':<15}")
print("-" * 80)

# 梯度下降迭代
for i in range(epochs + 1):
    # 1. 预测
    y_pred = w * x + b
    
    # 2. 计算损失 (MSE)
    loss = (1/n) * np.sum((y_pred - y)**2)
    
    # 3. 计算梯度
    dw = (2/n) * np.sum((y_pred - y) * x)
    db = (2/n) * np.sum(y_pred - y)
    
    # 记录日志
    if i in log_epochs:
        print(f"{i:<10} {w:<10.5f} {b:<10.5f} {loss:<15.2f} {dw:<15.1f} {db:<15.1f}")
    
    # 4. 更新参数
    w = w - lr * dw
    b = b - lr * db
```

## 3. 解析解验证

使用正规方程（Normal Equation）直接求解最优的 $w$ 和 $b$，以验证梯度下降的结果。

```python
# 计算均值
x_mean = np.mean(x)
y_mean = np.mean(y)

# 计算分子和分母
numerator = np.sum((x - x_mean) * (y - y_mean))
denominator = np.sum((x - x_mean)**2)

# 计算解析解
w_analytic = numerator / denominator
b_analytic = y_mean - w_analytic * x_mean

print("=== 解析解验证 ===")
print(f"解析解 w: {w_analytic}")
print(f"解析解 b: {b_analytic}")
print(f"梯度下降 w (第 {epochs} 轮): {w:.5f}")
print(f"梯度下降 b (第 {epochs} 轮): {b:.5f}")
```
