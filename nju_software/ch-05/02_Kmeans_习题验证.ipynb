# K-Means 习题验证 (6.1)

本 Notebook 用于验证文档中 6.1 节习题的手工推导过程。

```python
import numpy as np

# 数据点定义
points = {
    'A': np.array([2, 1]),
    'B': np.array([3, 5]),
    'C': np.array([6, 4]),
    'D': np.array([5, 6]),
    'E': np.array([7, 3]),
    'F': np.array([1, 3])
}

# 初始质心: A, C, E
# mu_1 = A, mu_2 = C, mu_3 = E
centroids = [points['A'].astype(float), points['C'].astype(float), points['E'].astype(float)]

print("初始质心:")
for idx, c in enumerate(centroids):
    print(f"mu_{idx+1}: {c}")
```

```python
prev_assignments = {}

for iteration in range(1, 5):
    print(f"\n#### 第 {iteration} 轮迭代")
    print("-" * 60)
    
    # Step 1: Assignment
    print("步骤 1：分配样本")
    current_assignments = {1: [], 2: [], 3: []}
    assignment_map = {}
    
    # Header
    print(f"{'点':<6} | {'到 mu1 d^2':<12} | {'到 mu2 d^2':<12} | {'到 mu3 d^2':<12} | {'归属'}")
    print("-" * 60)
    
    for name, p in sorted(points.items()):
        # Calculate squared Euclidean distance
        dists = [np.sum((p - c)**2) for c in centroids]
        
        # Find nearest centroid (1-based index)
        cluster_idx = np.argmin(dists) + 1 
        
        current_assignments[cluster_idx].append(name)
        assignment_map[name] = cluster_idx
        
        d_str = [f"{d:.2f}" for d in dists]
        print(f"{name:<6} | {d_str[0]:<12} | {d_str[1]:<12} | {d_str[2]:<12} | {cluster_idx}")

    # Step 2: Update Centroids
    print("\n步骤 2：更新质心")
    new_centroids = []
    for idx in range(1, 4):
        members = current_assignments[idx]
        if not members:
            new_c = centroids[idx-1] # Keep same if empty
            print(f"mu_{idx}' (无成员): {new_c}")
        else:
            coords = np.array([points[m] for m in members])
            new_c = np.mean(coords, axis=0)
            print(f"mu_{idx}' (成员 {members}): {new_c}")
        new_centroids.append(new_c)
    
    # Check convergence based on assignments
    if assignment_map == prev_assignments:
        print("\n观察：簇分配结果与上一轮一致。")
        print("算法收敛。")
        break
    
    prev_assignments = assignment_map
    centroids = new_centroids

print("\n最终结果：")
for idx in range(1, 4):
    print(f"簇 {idx}: {current_assignments[idx]}")
```
